Each assignment notebook consists of *cells*. There are different cell types:

* *Read-only* cells with the exercise description
* Editable code cells with the marker `# YOUR CODE HERE`. Write your solutions (as code) in these cells.
* *Read-only* cells with automatic tests. You can execute these cells to check if your solution passes the tests. Usually we will also have some additional *hidden tests* that are not visible to you, but will be executed after you hand in the assignment.
* Editable *markdown* cells with the marker `YOUR ANSWER HERE`. These cells are for written answers. You can use standard latex to write math symbols, e.g. `$f(x)=x^2$` will produce $f(x)=x^2$.

Depending on which program you use to work on the assignment (e.g. jupyter lab, VSCode, etc.), you may be able to edit the *read-only* cells. Keep in mind that the grading tool will replace these edited cells with the original cells after you submit your solution.
