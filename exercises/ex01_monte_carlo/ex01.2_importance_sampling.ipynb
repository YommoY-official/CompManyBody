## Exercise 1.2-b (2pt)

Implement the function `estimate_I_monte_carlo(rng, N, a, b=10)` that calculates $I$ by using that $I = \overline{f} \times (b-a) $ with $f(x) = \frac{\mathrm{e}^{-x}}{1+(x-1)^2} $ (You can find an estimate of $\overline{f}$ by sampling some $x_i \in [a,b)$ from an uniform distribution, and averaging over $f(x_i)$). Get an idea of the error of this estimate by plotting the mean error against the number of samples.

> **Note:** To get a value of $I$ to compare your estimate to you may use `scipy.integrate.quad`.


```python
def estimate_I_monte_carlo(rng, N, a, b=10):

    f_bar = 0
    for _ in range(N):
        f_bar += f(rng.uniform(a, b))
    f_bar /= N

    return f_bar * (b - a)
    # YOUR CODE HERE
```

```python
# 1.2-b tests
result_test = estimate_I_monte_carlo(np.random.default_rng(42), 1000, 0, 10)
assert np.isclose(result_test, 0.7, atol=2.e-1), f"The estimation for N=1000 samples should already be closer to the exact result."
del result_test
```

```python
# YOUR CODE HERE
#Compare mean error to number of samples
log_N_list = list(range(1,5))
N_bins = 10
rng = np.random.default_rng(42)
a = -3
b = 3

exact_I = scipy.integrate.quad(f,a,b)[0]

mean_error = []
for log_N in log_N_list:
    error = 0
    N = 10**log_N
    for _ in range(N_bins):
        I_est = estimate_I_monte_carlo(rng, N, a, b)
        error+= np.abs(I_est - exact_I)
    mean_error.append(np.log10(error/N_bins))

plt.plot(log_N_list, mean_error)
plt.title("Mean error of Monte Carlo Integral")
plt.xlabel("log10 Number of samples")
plt.ylabel("log10 Mean error")
plt.show()
```

Basically considering the integration as a rectangle where the height is the mean value of the function in that given range. could be a good estimate but there will defiinitely be functions that cannot be estimated with this. diverging functions for instance.

Implement the function `estimate_I_monte_carlo_importance_sampling(rng, N, alpha=1.46)`. Compare the estimation with and without importance sampling by recreating the plot from 1.2-a, now additionally including the sampling with importance sampling.


```python
def estimate_I_monte_carlo_importance_sampling(rng, N, alpha=1.46):
    # YOUR CODE HERE
    #assume b = 10 or more since the g(x) is practically 0 for values higher than 6 it seems
    sample_list = sample_from_pdf(rng, N, alpha)
    weights = f(sample_list) / g(sample_list, alpha)
    return np.mean(weights)
```

```python
# YOUR CODE HERE
rng = np.random.default_rng(42)
N = 10000
print(scipy.integrate.quad(f,0,10)[0])
print(estimate_I_monte_carlo(rng, N, 0, 10))
print(estimate_I_monte_carlo_importance_sampling(rng, N))
```
