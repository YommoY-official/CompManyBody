```python
def timeit(func_setup, func_eval, ns=[50, 100, 200, 400, 600, 1000, 2000, 4000, 6000, 10000], max_time=1.):
    print("starting run")
    ts = []
    for n in ns:
        print("  timing n={0}".format(n))
        args = func_setup(n)
        t0 = time.time()
        res = func_eval(*args)
        t1 = time.time()
        ts.append(t1-t0)
        if t1 - t0 > max_time:
            print(f"skip remaining since it took > {max_time:f}s")
            break
    return ns[:len(ts)], ts
```
